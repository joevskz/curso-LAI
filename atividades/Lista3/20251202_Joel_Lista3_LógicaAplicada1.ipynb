## QUESTÃO 1 - Gestão de Contas Bancárias
Implementação do diagrama UML fornecido, criando a classe Client e uma hierarquia de contas (Account, CheckingAccount, SavingsAccount)

```python
# ==============================================================================
# Defição da classe Client conforme o diagrama UML.
class Client():
    # Em Python, o método __init__ é o "construtor" que inicializa os atributos do objeto.
    def __init__(self, name, cpf, address, addressNumber, addressComplement, dtBirth, phone, email):
        # 'self' refere-se à instância atual do objeto sendo criado.
        self.name = name
        self.cpf = cpf
        self.address = address
        self.addressNumber = addressNumber
        self.addressComplement = addressComplement
        self.dtBirth = dtBirth
        self.phone = phone
        self.email = email

    # O método __str__ define como o objeto será apresentado se usarmos print(objeto).
    def __str__(self):
        return f"Cliente: {self.name} | CPF: {self.cpf} | Email: {self.email}"

    # Método para atualizar o telefone, retornando 1 para indicar sucesso (int)
    def updatePerson(self, new_phone):
        self.phone = new_phone
        print(f"Telefone de {self.name} atualizado para {self.phone}")
        return 1

# ==============================================================================
# Classe base Account (Conta Genérica)
class Account():
    def __init__(self, agency, agency_dig, code, client):
        self.agency = agency
        self.agency_dig = agency_dig
        self.code = code
        self.client = client  # Associação: A conta pertence a um objeto Client
        self.balance = 0.0    # Saldo inicializa zerado se não informado

    def update_balance(self):
        # Retorna o saldo atual
        return self.balance

# ==============================================================================
# Classe CheckingAccount (Conta Corrente) que HERDA de Account.
# A herança é indicada colocando a classe pai entre parênteses: class Filho(Pai).
class CheckingAccount(Account):
    def __init__(self, agency, agency_dig, code, client, account_number, account_dig, balance=0.0):
        # super().__init__ chama o construtor da classe pai (Account) para configurar os atributos básicos.
        super().__init__(agency, agency_dig, code, client)
        self.account_number = account_number
        self.account_dig = account_dig
        self.balance = balance  # Sobrescreve/define o saldo específico desta conta

    def __str__(self):
        return f"Conta Corrente: {self.account_number}-{self.account_dig} | Saldo: R$ {self.balance:.2f} | Titular: {self.client.name}"

    # Método para adicionar dinheiro (Depósito)
    def addMoney(self, value):
        self.balance += value
        print(f"Depósito de R$ {value:.2f} realizado na Conta Corrente.")
        return 1

    # Método para retirar dinheiro (Saque)
    def withdrawMoney(self, value):
        if self.balance >= value:
            self.balance -= value
            print(f"Saque de R$ {value:.2f} realizado com sucesso.")
            return 1
        else:
            print("Saldo insuficiente.")
            return 0

# ==============================================================================
# Classe SavingsAccount (Conta Poupança) que também HERDA de Account.
class SavingsAccount(Account):
    def __init__(self, agency, agency_dig, code, client, account_number, account_dig, balance=0.0):
        super().__init__(agency, agency_dig, code, client)
        self.account_number = account_number
        self.account_dig = account_dig
        self.balance = balance

    def __str__(self):
        return f"Conta Poupança: {self.account_number}-{self.account_dig} | Saldo: R$ {self.balance:.2f}"

    def addMoney(self, value):
        # Exemplo de lógica: Poupança pode ter um bônus ou regra diferente, mas aqui farei o básico.
        self.balance += value
        print(f"Depósito de R$ {value:.2f} realizado na Poupança.")
        return 1

    def withdrawMoney(self, value):
        if self.balance >= value:
            self.balance -= value
            print(f"Saque de R$ {value:.2f} da Poupança.")
            return 1
        else:
            print("Saldo insuficiente na Poupança.")
            return 0


# ==============================================================================
# --- TESTANDO O CÓDIGO (Instanciando Objetos) ---
print("--- TESTE QUESTÃO 1 ---")

print("-" * 65)
# 1. Criar um cliente
cliente1 = Client("João Silva", 12345678900, "Rua A", 10, "Apto 1", "1990-01-01", 99999999, "joao@email.com")
print(cliente1)
print("-" * 65)
# 2. Criar uma Conta Corrente para este cliente
cc = CheckingAccount(1, 1, 1001, cliente1, 5555, 0)
cc.addMoney(500.00)
cc.withdrawMoney(200.00)
print(cc)
print("-" * 65)
```

# Questão 2 - Criação da (Classe) Conta Salário na Gestão de Contas Bancárias
Adição da Classe SalaryAccount (Conta Salário) solicitada


```python
# =============================================================================
# Esta classe também estende Account, mas é uma classe especializada nova.
class SalaryAccount(Account):
    def __init__(self, agency, agency_dig, code, client, account_number, account_dig, balance=0.0):
        super().__init__(agency, agency_dig, code, client)
        self.account_number = account_number
        self.account_dig = account_dig
        self.balance = balance

    def __str__(self):
        return f"Conta Salário: {self.account_number}-{self.account_dig} | Saldo: R$ {self.balance:.2f}"

    def receive_salary(self, value):
        """Método específico de conta salário para receber pagamentos."""
        self.balance += value
        print(f"Salário de R$ {value:.2f} recebido.")
        return 1

    # Conta salário geralmente tem restrições de saque, mas implementei o básico
    def withdrawMoney(self, value):
        if self.balance >= value:
            self.balance -= value
            print(f"Saque de R$ {value:.2f} da Conta Salário.")
            return 1
        else:
            print("Saldo insuficiente.")
            return 0

# =============================================================================
# --- TESTANDO O CÓDIGO (Instanciando Objetos) ---
# Criar uma Conta Salário
cs = SalaryAccount(1, 1, 1002, cliente1, 7777, 9)

print("----- TESTE QUESTÃO 2 ------")

print("-" * 65)
print(cliente1)
print(cc)
cs.receive_salary(3000.00)
print(cs)
print("-" * 65)
```

# QUESTÃO 3 - Animais (Polimorfismo Básico)
Modelagem de uma classe base Animal e classes filhas Cachorro e Gato para demonstrar herança e polimorfismo (sobrescrita de métodos).

```python
# ==============================================================================
# Classe Base
class Animal():
    def __init__(self, nome, idade, som_caracteristico):
        self.nome = nome
        self.idade = idade
        self.som_caracteristico = som_caracteristico

    def exibir_informacoes(self):
        print(f"Animal: {self.nome}, Idade: {self.idade} anos")

    def emitir_som(self):
        # Este método será sobrescrito (polimorfismo) pelas subclasses
        print(f"O animal faz: {self.som_caracteristico}")

# =============================================================================
# Subclasse Cachorro
class Cachorro(Animal):
    def __init__(self, nome, idade):
        # Define o som fixo como "Latido" para todos os cachorros
        super().__init__(nome, idade, "Latido")

    # Sobrescrita de método (Polimorfismo):

    # O método existe na classe pai, mas o filho executa de forma diferente.
    def emitir_som(self):
        print(f"{self.nome} está latindo: Au Au!")

# =============================================================================
# Subclasse Gato
class Gato(Animal):
    def __init__(self, nome, idade):
        super().__init__(nome, idade, "Miado")

    def emitir_som(self):
        print(f"{self.nome} está miando: Miau!")

# =============================================================================
# --- TESTANDO O CÓDIGO ---
print("\n------ TESTE QUESTÃO 3 -------\n")
rex = Cachorro("Rex", 5)
mimi = Gato("Mimi", 3)

# Lista de animais para demonstrar polimorfismo
# O mesmo método 'emitir_som' se comporta diferente dependendo do objeto.
animais = [rex, mimi]
print("-" * 30)
for animal in animais:
    animal.exibir_informacoes()
    animal.emitir_som() # Polimorfismo em ação
    print("-" * 30)
```

# QUESTÃO 4 - Veículos (Herança e Implementação Específica)
Modelagem de Veiculo (base), Carro e Moto com o método mover() demonstrando comportamentos distintos.

```python
# =============================================================================
class Veiculo:
    def __init__(self, marca, modelo, ano):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano

    def exibir_informacoes(self):
        return f"Veículo: {self.marca} {self.modelo} ({self.ano})"

    def mover(self):
        print("O veículo está se movendo.")

# =============================================================================
class Carro(Veiculo):
    def mover(self):
        # Sobrescreve o comportamento genérico
        print(f"O carro {self.modelo} está rodando nas quatro rodas.")

# =============================================================================
class Moto(Veiculo):
    def mover(self):
        print(f"A moto {self.modelo} está rodando em duas rodas.")


# =============================================================================
# --- TESTANDO O CÓDIGO ---
print("\n----- TESTE QUESTÃO 4 -------")

print("-" * 60)
meu_carro = Carro("Fiat", "Strada", 2024)
minha_moto = Moto("Honda", "CG160", 2025)

print(meu_carro.exibir_informacoes())
meu_carro.mover()

print(minha_moto.exibir_informacoes())
minha_moto.mover()
print("-" * 60)
```

# QUESTÃO 5 - Televisores (Classes Abstratas e Interfaces)
Uso da biblioteca abc para criar classes que não podem ser instanciadas diretamente (Televisor e Conectividade) e obrigam as filhas a implementar seus métodos.

```python
# =============================================================================
from abc import ABC, abstractmethod

# =============================================================================
# Classe Abstrata Televisor
class Televisor(ABC): # Herdar de ABC (Abstract Base Class) torna esta classe abstrata.
    def __init__(self, marca, modelo, tamanho):
        self.marca = marca
        self.modelo = modelo
        self.tamanho = tamanho
        self.ligado = False
        self.volume = 10
        self.canal = 1

# =============================================================================
    # Método Concreto (já tem implementação e as filhas herdam direto)
    def exibir_informacoes(self):
        estado = "Ligada" if self.ligado else "Desligada"
        print(f"TV {self.marca} {self.modelo} - {self.tamanho} pol. | Estado: {estado}")

# =============================================================================
    # Métodos Abstratos (Funcionam como uma 'regra': as filhas SÃO OBRIGADAS a implementar)
    @abstractmethod
    def ligar(self):
        pass

    @abstractmethod
    def desligar(self):
        pass

# =============================================================================
    # Métodos de controle de canal e volume solicitados na lista
    @abstractmethod
    def aumentar_volume(self):
        pass

    @abstractmethod
    def diminuir_volume(self):
        pass

# =============================================================================
# Interface Conectividade (Simulada com Classe Abstrata)
# Define um contrato que diz: "Quem herdar de mim, tem que saber conectar na internet"
class Conectividade(ABC):
    @abstractmethod
    def conectar_internet(self):
        pass

# =============================================================================
# Classe Concreta SmartTV
# Herança Múltipla: Herda de Televisor E de Conectividade
class SmartTV(Televisor, Conectividade):
    def ligar(self):
        self.ligado = True
        print("Ligando a Smart TV com sistema operacional Android TV...")

    def desligar(self):
        self.ligado = False
        print("Encerrando apps e desligando Smart TV.")

    def aumentar_volume(self):
        self.volume += 1
        print(f"Volume SmartTV: {self.volume}")

    def diminuir_volume(self):
        self.volume -= 1
        print(f"Volume SmartTV: {self.volume}")

    # Implementação obrigatória da interface Conectividade
    def conectar_internet(self):
        print("Conectando via Wi-Fi... Sucesso! Acesso ao HBO-MAX liberado.")

# =============================================================================
# Classe Concreta TVComum
# A lista pede para implementar a interface, mesmo que não tenha internet de fato.
class TVComum(Televisor, Conectividade):
    def ligar(self):
        self.ligado = True
        print("Ligando a TV comum com controle remoto padrão...")

    def desligar(self):
        self.ligado = False
        print("Desligando TV comum.")

    def aumentar_volume(self):
        self.volume += 1
        print(f"Volume TV Comum: {self.volume}")

    def diminuir_volume(self):
        self.volume -= 1
        print(f"Volume TV Comum: {self.volume}")

    def conectar_internet(self):
        # A implementação aqui é apenas informativa, conforme solicitado no enunciado da atividade
        print("Esta TV não possui conexão com a internet.")

# =============================================================================
# --- TESTANDO O CÓDIGO ---
print("\n----- TESTE QUESTÃO 5 ------")

print("-" * 60)
# Instanciando SmartTV
minha_smart = SmartTV("TCL", "55C6K", 55)
minha_smart.exibir_informacoes()
minha_smart.ligar()
minha_smart.conectar_internet()
minha_smart.aumentar_volume()

print("-" * 60)

# Instanciando TV Comum
minha_tubo = TVComum("LG", "32LT330HBSB", 32)
minha_tubo.exibir_informacoes()
minha_tubo.ligar()
minha_tubo.conectar_internet() # Vai avisar que não tem net
print("-" * 60)
```
