## 1. Criar um algoritmo para ler os lados de um triângulo e identificar se é equilátero.

```python
# Questão 1: Identificar se o triângulo é equilátero

# 1. Pede os 3 lados e tenta transformar em número
try:
    lado1 = float(input("Digite o primeiro lado: "))
    lado2 = float(input("Digite o segundo lado: "))
    lado3 = float(input("Digite o terceiro lado: "))
    
    # 2. Valida se os lados são positivos
    if lado1 > 0 and lado2 > 0 and lado3 > 0:
        # 3. Se forem positivos, verifica se são equiláteros
        if lado1 == lado2 and lado2 == lado3:
            print("É um triângulo equilátero!")
        else:
            print("Não é um triângulo equilátero.")
    else:
        print("Os lados do triângulo precisam ser números positivos.")

except ValueError:
    # 4. Caso o usuário digite algo que não seja um número
    print("Entrada inválida. Por favor, digite apenas números.")
```

## 3. Adapte o algoritmo anterior para identificar o maior lado de um triângulo escaleno.

```python
# Questão 3: Encontrar o maior lado de um triângulo escaleno

# 1. Pede para o usuário digitar os 3 lados
lado1 = float(input("Digite o primeiro lado: "))
lado2 = float(input("Digite o segundo lado: "))
lado3 = float(input("Digite o terceiro lado: "))

# 2. Verifica se o triângulo é escaleno (todos os lados diferentes)
if lado1 != lado2 and lado1 != lado3 and lado2 != lado3:
    print("O triângulo é escaleno.")
    
    # 3. Encontra o maior lado usando comparações
    maior_lado = lado1
    if lado2 > maior_lado:
        maior_lado = lado2
    if lado3 > maior_lado:
        maior_lado = lado3
    
    print("O maior lado é:", maior_lado)
else:
    print("O triângulo não é escaleno, então não posso dizer qual é o maior lado.")
```

## 4. Faça um algoritmo que calcule a área de um Triângulo (os valores da base e altura devem ser lidos do teclado).

```python
# Questão 4: Calcular a área de um triângulo

# 1. Pede a base e a altura e tenta transformar em número
try:
    base = float(input("Digite o valor da base: "))
    altura = float(input("Digite o valor da altura: "))

    # 2. Valida se a base e a altura são positivas
    if base > 0 and altura > 0:
        # 3. Se forem positivas, calcula a área
        area = (base * altura) / 2
        # 4. Mostra o resultado
        print("A área do triângulo é:", area)
    else:
        print("A base e a altura precisam ser números positivos.")

except ValueError:
    # 5. Caso o usuário digite algo que não seja um número
    print("Entrada inválida. Por favor, digite apenas números.")
```

## 5. Faça um algoritmo que leia o preço de um produto e mostre seu novo preço com 25% de desconto.

```python
# Questão 5: Calcular preço com 25% de desconto

# 1. Pede o preço do produto
preco_original = float(input("Digite o preço do produto: R$ "))

# 2. Calcula o desconto e o novo preço
valor_desconto = preco_original * 0.25
novo_preco = preco_original - valor_desconto

# 3. Mostra o resultado de forma simples
print("O novo preço com desconto é:", novo_preco)
```

## 6. Faça um algoritmo que leia a idade de uma pessoa expressa em anos, meses e dias e escreva a idade dessa pessoa em dias. Considerar ano com 365 dias e mês com 30 dias.

```python
# Questão 6: Converter idade para dias

# 1. Pede a idade em anos, meses e dias
anos = int(input("Digite a quantidade de anos: "))
meses = int(input("Digite a quantidade de meses: "))
dias = int(input("Digite a quantidade de dias: "))

# 2. Faz as contas
total_dias = (anos * 365) + (meses * 30) + dias

# 3. Mostra o resultado
print("A idade total em dias é:", total_dias)
```
